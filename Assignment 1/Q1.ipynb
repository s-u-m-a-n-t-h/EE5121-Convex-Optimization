```python
#imports
import cvxpy as cp
import numpy as np
```

```python
#Constant matrix A, vector b and 1-vector
A = np.array([[1,0,1],[0,1,1],[1,1,2]])
b = np.array([0.2,0.8,1])
one = np.array([1,1,1])
```

```python
#Decision Variables
x = cp.Variable(3)
u = cp.Variable(3)
```

(d)

```python
#Formulating the problem
objective = cp.Minimize(one.T@u)
constraints = [A@x == b, -u <= x, x <= u, u >= 0]
prob = cp.Problem(objective=objective,constraints=constraints)
```

```python
#minimum and the optimizer
print("Optimal objective value", prob.solve())
print("Optimal variable value")
print(x.value)
```

```python
print(u.value)
```

(e)

```python
#different device
b = np.array([0.2,0.8,0.95])
```

```python
#reformulating problem 
objective = cp.Minimize(one.T@u + 1*cp.norm(A@x - b))
constraints = [-u <= x, x <= u, u >= 0]
prob = cp.Problem(objective=objective,constraints=constraints)
```

```python
#minimum and the optimizer
print("Optimal objective value", prob.solve())
print("Optimal variable value")
print(x.value)
```

```python
print(u.value)
```

```python
#residual 
print(np.linalg.norm(A@x.value - b))
```
