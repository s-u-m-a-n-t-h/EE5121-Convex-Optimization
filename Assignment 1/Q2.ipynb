```python
#imports
import cvxpy as cp
import numpy as np
```

```python
#input and measurement constant matrices
A = np.array([[1,0,0,0,0],[0.8,1,0,0,0],[-0.2,0.8,1,0,0],\
              [0.5,-0.2,0.8,1,0],[0,0.5,-0.2,0.8,1]])

B = np.array([[1,-1,0,0,1],[2.8,-0.8,1,-1,0.8],[2.4,1.2,1.8,-1.8,-1.2]\
              ,[2.9,-1.7,0.6,-0.6,1.7],[0.4,-1.8,-0.7,0.7,1.8]])
```

```python
#decision variable
X = cp.Variable((5,5))
```

(c) 1)

```python
#formulation
objective = cp.Minimize(cp.normNuc(X))
constraints = [A@X == B]
problem = cp.Problem(objective=objective,constraints=constraints)
```

```python
#minimum and the optimizer
print("Optimal objective value", problem.solve())
print("Optimal variable value")
print(X.value)
```

```python
#rounnding-off
np.round(X.value,0)
```

```python
#rank - to see if the nuclear norm minimization is achieving the desired effect
np.linalg.matrix_rank(np.round(X.value,0))
```

```python
#Gaussian Noise matrix for i = j
n = [np.random.normal(0,np.sqrt(0.1)) for i in range(5)]
N = np.zeros((5,5))
for i in range(5):
    N[i,i] = n[i]
```

```python
# Corrupted B
B = B + N
```

```python
#reformulation
objective = cp.Minimize(cp.normNuc(X) + cp.norm(A@X - B,"fro"))
problem = cp.Problem(objective=objective,constraints=constraints)
```

```python
#minimum and minimizer
print("Optimal objective value", problem.solve())
print("Optimal variable value")
print(X.value)
```
